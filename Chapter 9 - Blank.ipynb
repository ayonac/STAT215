```python
#Creating the Flip function
# Here x is the input you give, and a_b is the array that stores
#the two possible values of the input
def flip_func(x, a_b):
    if x==a_b.item(0):
        return a_b.item(1)
    elif x==a_b.item(1):
        return a_b.item(0)
    else:
        return 'Input not valid'
```

```python
#Lets run the above function to check
pets=make_array('Dog','Cat')
pets
```
